```python
def fun(x):
    return (x+3)**2

def derivative(x):
    return 2 * (x + 3)

x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    grad = derivative(x)
    x = x - learning_rate * grad

print("Local minimum at x = ", round(x, 4))
print("Minimum value of y = ", round(fun(x), 4))
```
